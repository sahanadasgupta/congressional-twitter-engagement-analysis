```python
import pandas as pd

# Reload users_df to ensure original data for robust parsing
users_df = pd.read_csv('/content/drive/MyDrive/users_data_flattened.csv')

def parse_created_at_robust(date_string):
    if pd.isna(date_string): # Handle NaN values if any
        return pd.NaT
    date_string = str(date_string) # Ensure it's a string

    # Attempt to parse as Unix timestamp (numeric string)
    if date_string.isdigit():
        try:
            return pd.to_datetime(int(date_string), unit='s', errors='coerce', utc=True)
        except (ValueError, TypeError):
            pass # Fallback to mixed format if int conversion fails unexpectedly

    # Attempt to parse as mixed format date string
    return pd.to_datetime(date_string, format='mixed', errors='coerce', utc=True)

users_df['created_at'] = users_df['created_at'].apply(parse_created_at_robust)

# Reload tweets_df for consistency and convert its 'created_at' column (which is int64 Unix timestamp)
tweets_df = pd.read_csv('/content/drive/MyDrive/tweets_data.csv')
tweets_df['created_at'] = pd.to_datetime(tweets_df['created_at'], unit='s', errors='coerce')

print('--- Users DataFrame created_at Dtype ---\n')
print(users_df['created_at'].dtype)
print('\n--- Tweets DataFrame created_at Dtype ---\n')
print(tweets_df['created_at'].dtype)

# Check for duplicate user IDs in users_df
duplicate_user_ids = users_df[users_df.duplicated(subset=['id'])]

print('\n--- Duplicate User IDs in users_df ---\n')
if not duplicate_user_ids.empty:
    print(f"Found {len(duplicate_user_ids)} duplicate user IDs.")
    display(duplicate_user_ids.head())
else:
    print("No duplicate user IDs found.")

print('\n--- Number of NaT values in users_df["created_at"] after fix ---\n')
print(users_df['created_at'].isnull().sum())
```

## Proposed Entity-Relationship Diagram (ERD) Description

Based on the `users_df` and `tweets_df` datasets, we can propose an ERD with two primary entities: **User** and **Tweet**. These entities capture the core information about users on the platform and the tweets they create. Below is a detailed description of each entity, their key attributes, and the relationships between them.

### Entities and Their Attributes

#### 1. User

This entity represents individual user profiles, derived from the `users_df`.

*   **Primary Key:**
    *   `id` (Unique identifier for each user)

*   **Key Attributes:**
    *   `screen_name` (User's unique Twitter handle)
    *   `name` (User's display name)
    *   `created_at` (Date and time when the user account was created)
    *   `description` (User's self-provided description/bio)
    *   `location` (User's reported location)
    *   `followers_count` (Number of followers the user has)
    *   `friends_count` (Number of users the user is following)
    *   `statuses_count` (Total number of tweets posted by the user)
    *   `favourites_count` (Total number of tweets the user has liked)
    *   `listed_count` (Number of times the user has been added to a public list)
    *   `verified` (Boolean indicating if the user is verified)
    *   `lang` (Language setting of the user's account)
    *   `profile_image_url_https` (URL to the user's profile picture)
    *   `url` (User's website URL, if provided)

#### 2. Tweet

This entity represents individual tweets posted on the platform, derived from the `tweets_df`.

*   **Primary Key:**
    *   `id` (Unique identifier for each tweet)

*   **Key Attributes:**
    *   `user_id` (Foreign Key referencing `User.id` – the ID of the user who posted the tweet)
    *   `created_at` (Date and time when the tweet was posted)
    *   `text` (The actual content of the tweet)
    *   `source` (The client application used to post the tweet, e.g., 'Twitter for iPhone')
    *   `favorite_count` (Number of likes the tweet has received)
    *   `retweet_count` (Number of times the tweet has been retweeted)
    *   `lang` (Language detected for the tweet content)
    *   `is_quote_status` (Boolean indicating if the tweet is a quote tweet)
    *   `in_reply_to_status_id` (ID of the tweet this tweet is a reply to, if applicable)
    *   `in_reply_to_user_id` (ID of the user this tweet is a reply to, if applicable)
    *   `quoted_status_id` (ID of the tweet being quoted, if applicable)
    *   `entities.hashtags` (List of hashtags used in the tweet)
    *   `entities.user_mentions` (List of user mentions in the tweet)
    *   `entities.urls` (List of URLs included in the tweet)
    *   `possibly_sensitive` (Boolean indicating if the tweet is marked as potentially sensitive)

### Relationships

#### 1. `User` posts `Tweet`

*   **Type:** One-to-Many
*   **Cardinality:** `User` (1) ------ `Tweets` (Many)
*   **Description:** A single `User` can post multiple `Tweets`, but each `Tweet` is posted by exactly one `User`.
*   **Linkage:** `Tweet.user_id` is a Foreign Key that references `User.id`.

#### 2. `Tweet` replies to/quotes `Tweet` (Self-referencing)

*   **Type:** Many-to-One (for replies/quotes)
*   **Cardinality:** `Tweet` (Many) ------ `Tweet` (1) (A reply/quote references one original tweet)
*   **Description:** A `Tweet` can be a reply to another `Tweet` or quote another `Tweet`. This is a recursive relationship where `Tweet` instances are related to other `Tweet` instances.
*   **Linkage:**
    *   `Tweet.in_reply_to_status_id` is a Foreign Key that references `Tweet.id` (for replies).
    *   `Tweet.quoted_status_id` is a Foreign Key that references `Tweet.id` (for quote tweets).

### Visual Representation Guidelines for ERD

When drawing the ERD, you would represent each entity (User, Tweet) as a rectangle. Attributes would be listed within or below the entity rectangle, with primary keys underlined. Relationships would be depicted as lines connecting the entities, with crows' foot notation (or similar) to indicate cardinality (e.g., a straight line on the 'one' side and a three-pronged foot on the 'many' side).

This ERD provides a clear structure for understanding the data and can guide further analysis and potential database design.

```python
import pandas as pd
import numpy as np
import re
from sklearn.linear_model import LinearRegression
from sklearn.feature_extraction.text import TfidfVectorizer

# Ensure datetime and merged datasets are ready
# Ensure datetime and merged datasets are ready
tweets_df['created_at'] = pd.to_datetime(tweets_df['created_at'], unit='s', errors='coerce')
merged_df = tweets_df.merge(users_df, left_on='user_id', right_on='id', suffixes=('', '_user'))
# ---------------------------------------------------------
# 1. DIVE DEEPER: CORRELATIONS & LINEAR REGRESSION
# ---------------------------------------------------------
print('=== Pearson & Spearman Correlations ===')
# Calculate Pearson Correlation
pearson_corr = merged_df[['followers_count', 'friends_count', 'retweet_count', 'favorite_count']].corr(method='pearson')
display(pearson_corr)

# Linear Regression: Predict Retweets based on Followers
# Filter out missing values for modeling
reg_data = merged_df[['followers_count', 'retweet_count']].dropna()
X = reg_data[['followers_count']]
y = reg_data['retweet_count']

model = LinearRegression()
model.fit(X, y)

print(f"\nLinear Regression Model:")
print(f"Slope (Coefficient): {model.coef_[0]:.6f}")
print(f"Intercept: {model.intercept_:.2f}")
print(f"R-squared: {model.score(X, y):.4f}")

# ---------------------------------------------------------
# 2. DIVE DEEPER: TEXTUAL ANALYSIS (TF-IDF)
# ---------------------------------------------------------
print('\n=== TF-IDF Text Analysis ===')
# Basic text cleaning function
def clean_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)  # Remove URLs
    text = re.sub(r'\@\w+|\#','', text)  # Remove mentions and hashtag symbols
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # Remove special chars/numbers
    return text

# Clean tweet text
tweets_df['cleaned_text'] = tweets_df['text'].apply(clean_text)

# Run TF-IDF Vectorizer with English stop-words removed
tfidf = TfidfVectorizer(max_features=15, stop_words='english')
tfidf_matrix = tfidf.fit_transform(tweets_df['cleaned_text'].dropna())

# Display Top TF-IDF Terms
feature_names = tfidf.get_feature_names_out()
avg_tfidf_scores = tfidf_matrix.mean(axis=0).A1
top_terms = pd.DataFrame({'Term': feature_names, 'Mean TF-IDF Score': avg_tfidf_scores})
top_terms = top_terms.sort_values(by='Mean TF-IDF Score', ascending=False)
display(top_terms)

# ---------------------------------------------------------
# 3. GO BROADER: EXPANDED FEATURE CONNECTIONS
# ---------------------------------------------------------
print('\n=== Broader Feature Exploration ===')
# Feature: Verification Status vs Engagement
verified_stats = merged_df.groupby('verified')[['retweet_count', 'favorite_count']].agg(['count', 'mean', 'median'])
display(verified_stats)

# Feature: Has Links vs Engagement
merged_df['has_url'] = merged_df['entities.urls'].notna()
url_stats = merged_df.groupby('has_url')[['retweet_count', 'favorite_count']].agg(['count', 'mean', 'median'])
display(url_stats)

# ---------------------------------------------------------
# 4. NEW METRICS
# ---------------------------------------------------------
print('\n=== New Custom Metrics ===')
# Metric 1: Engagement Rate per Follower (Ratio of total engagements to user followers)
# Adding 1 to denominator to avoid division by zero
merged_df['engagement_per_follower'] = (merged_df['retweet_count'] + merged_df['favorite_count']) / (merged_df['followers_count'] + 1)

# Metric 2: Amplification Efficiency Ratio (Ratio of retweets to favorites)
# Higher ratio indicates content that spreads beyond existing fan base
merged_df['amplification_ratio'] = merged_df['retweet_count'] / (merged_df['favorite_count'] + 1)

# Display summary of new metrics
display(merged_df[['engagement_per_follower', 'amplification_ratio']].describe())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set overall aesthetic style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12})

# Ensure dataset is loaded and preprocessed
users_df = pd.read_csv('/content/drive/MyDrive/users_data_flattened.csv')
tweets_df = pd.read_csv('/content/drive/MyDrive/tweets_data.csv')

tweets_df['has_media'] = tweets_df['entities.media'].notna()
tweets_df['created_at'] = pd.to_datetime(tweets_df['created_at'], unit='s', errors='coerce')
tweets_df['created_hour'] = tweets_df['created_at'].dt.hour
merged_df = tweets_df.merge(users_df, left_on='user_id', right_on='id', suffixes=('', '_user'))

# ---------------------------------------------------------
# 1. FORMAL HYPOTHESIS TESTING (Statistical Rigor)
# ---------------------------------------------------------
print("=== FORMAL STATISTICAL HYPOTHESIS TESTS ===\n")

# Test 1: Two-Sample Welch's T-Test (Media vs Plain Text Retweets)
media_retweets = tweets_df[tweets_df['has_media'] == True]['retweet_count'].dropna()
plain_retweets = tweets_df[tweets_df['has_media'] == False]['retweet_count'].dropna()
t_stat, p_val_media = stats.ttest_ind(media_retweets, plain_retweets, equal_var=False)

print(f"Hypothesis 1 (Media Effect - Retweets):")
print(f"  t-statistic = {t_stat:.4f}, p-value = {p_val_media:.4e}")
print(f"  Result: {'Statistically Significant (Reject Null)' if p_val_media < 0.05 else 'Not Significant'}\n")

# Test 2: One-Way ANOVA (Engagement across Peak vs Off-Peak Hours)
hourly_groups = [group['retweet_count'].dropna() for name, group in tweets_df.groupby('created_hour')]
f_stat, p_val_hours = stats.f_oneway(*hourly_groups)

print(f"Hypothesis 2 (Hourly Variation):")
print(f"  F-statistic = {f_stat:.4f}, p-value = {p_val_hours:.4e}")
print(f"  Result: {'Statistically Significant Differences Exist' if p_val_hours < 0.05 else 'Not Significant'}\n")

# Test 3: Spearman Correlation Test (Followers vs Engagement)
spearman_r, p_val_followers = stats.spearmanr(merged_df['followers_count'], merged_df['favorite_count'], nan_policy='omit')

print(f"Hypothesis 3 (Follower Impact on Favorites):")
print(f"  Spearman rho = {spearman_r:.4f}, p-value = {p_val_followers:.4e}")
print(f"  Result: {'Statistically Significant Positive Correlation' if p_val_followers < 0.05 else 'Not Significant'}\n")

# ---------------------------------------------------------
# 2. GENERATE VISUALIZATIONS FOR SLIDES
# ---------------------------------------------------------

# Figure 1: Media vs Plain Text Engagement
plt.figure(figsize=(8, 5))
sns.barplot(data=tweets_df, x='has_media', y='retweet_count', palette='Blues_d', ci=None)
plt.title('Average Retweet Count: Media vs. Plain Text Tweets', fontsize=14, fontweight='bold')
plt.xlabel('Has Media Attachment', fontsize=12)
plt.ylabel('Mean Retweet Count', fontsize=12)
plt.xticks([0, 1], ['Plain Text (False)', 'Contains Media (True)'])
plt.tight_layout()
plt.savefig('fig1_media_engagement.png', dpi=300)
plt.show()

# Figure 2: Tweet Volume vs. Average Engagement by Hour
hourly_agg = tweets_df.groupby('created_hour').agg(
    volume=('id', 'count'),
    mean_retweets=('retweet_count', 'mean')
).reset_index()

fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()

sns.barplot(data=hourly_agg, x='created_hour', y='volume', ax=ax1, color='lightblue', alpha=0.6)
sns.lineplot(data=hourly_agg, x='created_hour', y='mean_retweets', ax=ax2, color='darkblue', marker='o', linewidth=2.5)

ax1.set_xlabel('Hour of Day (UTC)', fontsize=12)
ax1.set_ylabel('Total Tweet Volume (Bars)', color='blue', fontsize=12)
ax2.set_ylabel('Mean Retweet Count (Line)', color='darkblue', fontsize=12)
plt.title('Posting Volume vs. Average Engagement Rate by Hour', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('fig2_hourly_mismatch.png', dpi=300)
plt.show()

# Figure 3: Engagement by Follower Tier
merged_df['follower_tier'] = pd.qcut(merged_df['followers_count'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])
tier_agg = merged_df.groupby('follower_tier', observed=False)['favorite_count'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=tier_agg, x='follower_tier', y='favorite_count', palette='Purples_d')
plt.title('Mean Favorite Count across User Follower Tiers', fontsize=14, fontweight='bold')
plt.xlabel('Follower Tier Quantile', fontsize=12)
plt.ylabel('Mean Favorite Count', fontsize=12)
plt.tight_layout()
plt.savefig('fig3_follower_tiers.png', dpi=300)
plt.show()
```
